01 Exercício : Desenvova um programa que seecione um numero aleatorio entre 0 e 100.O usuario devera tentrar adicionar ese numero.Ao final o programa devera exibir a sequencia de papites do usuario e o total de fatas ate acerta o numero correto 

```python
import random

def advinha_numero():
    numero_secreto = random.randint(0, 100)
    tentativas = []
    acertou = False

    while not acertou:
        palpite = int(input("Digite seu palpite: "))
        tentativas.append(palpite)

        if palpite == numero_secreto:
            acertou = True
            print("Parabéns! Você acertou o número!!!")
        elif palpite < numero_secreto:
            print("O número é maior que esse!")
        else:
            print("O número é menor que esse!")

    print(f"Você acertou o número {numero_secreto} em {len(tentativas)} tentativas.")
    print("Tentativas de palpites: ", tentativas)

# Chama a função para iniciar o jogo
advinha_numero()
```
