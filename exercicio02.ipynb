Exercício 2: Você esta organizando uma lista de alunos por turma por uma escola .Cada turma  ocntem varios alunos  e cada aluno é representaodr por um dicionario  com informações sobre seu nome e idade 

a) Crie  as turmas Crie duas turmas ,cada uma contendo uma lista de dicionarios representando os alunos .Cada dicionario dev e ter as as chaves "nome" e "idade"

b) Adicionar um novo aluno:Adiicione um noovo aluno em cada uma das turmas 

c) Remover um aluno : remova um aluno de uma das turmas

d) Exibir os alunos de uma das turmas 

```python
 #Criando as turmas 
turma1 = [
    {"nome": "Ana", "idade": 12},
    {"nome": "Bruno", "idade": 11}
]

turma2 = [
    {"nome": "Carla", "idade": 13},
    {"nome": "Daniel", "idade": 12}
]

#adicionar um novo aluno 
novo_aluno_turma1 = {"nome": "Eduarda", "idade": 11}
novo_aluno_turma2 = {"nome": "Fábio", "idade": 13}

turma1.append(novo_aluno_turma1)
turma2.append(novo_aluno_turma2)

# remover aluno
turma1.pop(0)

#exibir os alunos de uma das turmas 
print("Alunos da Turma 1:")
for aluno in turma1:
    print(f"Nome: {aluno['nome']}, Idade: {aluno['idade']}")
```
