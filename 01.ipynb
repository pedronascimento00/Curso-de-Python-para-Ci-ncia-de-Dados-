### Estrutura de Decisão
- Só vai fazer algo se a condição  for verdadeira 

se {acabou café}

    fazer mais café

senão

    beber café

```python
# exemplo 

x:int = int(input("Adcione um numero : "))

if (x > 10):
    print(f"{x} é um numero maior que 10")
else:
    print(f"{x} é menor ou igual a 10" )
print("Essa mensagem não depende de x ")
```

### While


```python
#while(condição)
i :int = 10

while(i < 10):
    print(i)
    time.sleep(i)

    i = i -1
```

### For 

```python

for i in range(a,b,c):
    print(i,end ="")
```

### Break e Continue 

```python
i: int = 0
while(True):
    if(i < 10):
        i += 1

        continue
        print(i)
        break
```
